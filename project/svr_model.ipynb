```python
# Support Vector Regression (SVR)

## Objective

# Can we predict an actor's billing position in a Marvel production
# based on their previous appearances and character history?

# This notebook implements Support Vector Regression (SVR) to predict
# actor billing positions using the previously engineered features.
```

```python
%pip install pandas numpy matplotlib scikit-learn
from sklearn.dummy import DummyRegressor
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
train = pd.read_csv("data_processed/train_scaled.csv")
test = pd.read_csv("data_processed/test_scaled.csv")

print("Training shape:", train.shape)
print("Testing shape:", test.shape)

train.head()
```

```python
X_train = train.drop(columns=["billing_tier", "billing_position"])
y_train = train["billing_position"]

X_test = test.drop(columns=["billing_tier", "billing_position"])
y_test = test["billing_position"]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
baseline = DummyRegressor(strategy="median")

baseline.fit(X_train, y_train)

baseline_predictions = baseline.predict(X_test)

baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_predictions))
baseline_r2 = r2_score(y_test, baseline_predictions)

print("Baseline Model")
print("-" * 30)
print(f"MAE:  {baseline_mae:.4f}")
print(f"RMSE: {baseline_rmse:.4f}")
print(f"R²:   {baseline_r2:.4f}")
```
